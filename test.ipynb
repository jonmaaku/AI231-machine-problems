```python
# Given a list of request latencies in ms and a parallel list of tokens generated per request, 
# compute p50/p95/p99 and average tokens/sec (sum(tokens)/sum(latency_s)).

import numpy as np

def compute_latency_metrics(latencies_ms, tokens_per_request):
    """
    Compute latency percentiles and average tokens per second.
    
    Args:
        latencies_ms: List of request latencies in milliseconds
        tokens_per_request: List of tokens generated per request (parallel to latencies)
    
    Returns:
        Dictionary containing p50, p95, p99 latencies and average tokens/sec
    """
    # Convert to numpy arrays for easier computation
    latencies = np.array(latencies_ms)
    tokens = np.array(tokens_per_request)
    
    # Compute percentiles (p50, p95, p99)
    p50 = np.percentile(latencies, 50)
    p95 = np.percentile(latencies, 95)
    p99 = np.percentile(latencies, 99)
    
    # Convert latencies to seconds and compute average tokens/sec
    latencies_seconds = latencies / 1000.0  # Convert ms to seconds
    total_tokens = np.sum(tokens)
    total_time_seconds = np.sum(latencies_seconds)
    avg_tokens_per_sec = total_tokens / total_time_seconds if total_time_seconds > 0 else 0
    
    return {
        'p50_ms': p50,
        'p95_ms': p95,
        'p99_ms': p99,
        'avg_tokens_per_sec': avg_tokens_per_sec,
        'total_requests': len(latencies),
        'total_tokens': total_tokens,
        'total_time_sec': total_time_seconds
    }

# Example usage:
# Sample data
latencies_ms = [120, 85, 200, 150, 95, 300, 180, 110, 250, 90, 400, 160, 75, 350, 140]
tokens_per_request = [50, 30, 80, 60, 40, 120, 70, 45, 90, 35, 150, 65, 25, 110, 55]

# Compute metrics
results = compute_latency_metrics(latencies_ms, tokens_per_request)

# Display results
print("Latency Metrics:")
print(f"P50 (median): {results['p50_ms']:.2f} ms")
print(f"P95: {results['p95_ms']:.2f} ms") 
print(f"P99: {results['p99_ms']:.2f} ms")
print(f"\nThroughput:")
print(f"Average tokens/sec: {results['avg_tokens_per_sec']:.2f}")
print(f"\nSummary:")
print(f"Total requests: {results['total_requests']}")
print(f"Total tokens: {results['total_tokens']}")
print(f"Total time: {results['total_time_sec']:.3f} seconds")
```
